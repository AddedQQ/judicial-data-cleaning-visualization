```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path(r"C:\Users\Luciano Sotelo\Documents\TESIS_PODER_JUDICIAL")

RUTA_BASE_ENRIQUECIDA = BASE_DIR / "02_datos_limpios" / "base_poder_judicial_enriquecida_v1.xlsx"
RUTA_POWERBI = BASE_DIR / "05_powerbi_dashboard" / "base_powerbi_v1.xlsx"

RUTA_POWERBI.parent.mkdir(exist_ok=True)

df = pd.read_excel(RUTA_BASE_ENRIQUECIDA)

print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

df.head()
```

```python
df = df.copy()

df["ID_Caso"] = range(1, len(df) + 1)

df[["ID_Caso", "Fecha Intervención", "Año", "Mes"]].head()
```

```python
variables_violencia = [
    "Violencia Física",
    "Violencia Psicológica",
    "Violencia Económica",
    "Violencia Sexual",
    "Violencia Restricción Participación Social",
    "Violencia Ambiental"
]

medidas_judiciales = [
    "Abstención de publicación en Redes Sociales",
    "Alojamiento temporario de la víctima",
    "Custodia policial",
    "Desestimación de denuncia",
    "Exclusión del agresor",
    "Fijación de cuota alimentaria provisoria",
    "Fijación de régimen de visitas",
    "Guarda de menores/incapaces",
    "Guarda de menores/incapaces víctimas",
    "Levantamiento de la medida judicial",
    "Oficio de protección policial",
    "Orden para retirar enseres personales",
    "Otras Medidas",
    "Prohibición de acercamiento o comunicación del agresor",
    "Recursos de Revocatoria",
    "Reintegro al hogar a las víctimas",
    "Restitución de alimentos",
    "Restitución de bienes",
    "Retiro de la víctima",
    "Secuestro de arma",
    "Tratamientos de rehabilitación"
]
```

```python
mapa_gravedad = {
    0: "No constatada",
    1: "Baja",
    2: "Relativa",
    3: "Alta",
    4: "Muy alta"
}

mapa_relacion = {
    1: "Unión convivencial",
    2: "Cónyuge",
    3: "Filial",
    4: "Fraternal",
    5: "Otros",
    6: "Novios",
    7: "Ex pareja",
    8: "Otros parientes"
}

mapa_frecuencia = {
    1: "Primer episodio",
    2: "Esporádico",
    3: "Mensual",
    4: "Quincenal",
    5: "Semanal",
    6: "Diario"
}

mapa_nivel_educativo = {
    1: "Analfabeta/o",
    2: "Primario incompleto",
    3: "Primario completo",
    4: "Secundario incompleto",
    5: "Secundario completo",
    6: "Terciario incompleto",
    7: "Terciario completo",
    8: "Universitario incompleto",
    9: "Universitario completo",
    10: "Nivel inicial - Jardín",
    11: "Post universitaria"
}

mapa_condicion_laboral = {
    1: "Ocupada remunerada",
    2: "Desocupada",
    3: "Jubilada/Pensionada",
    4: "Trabaja sin remuneración/Ama de casa",
    5: "Otra ocupación",
    6: "Infancia"
}
```

```python
columnas_casos = [
    "ID_Caso",
    "Fecha Intervención",
    "Año",
    "Mes",
    "Sexo denunciante",
    "Edad denunciante limpia",
    "Grupo etario denunciante",
    "Sexo denunciada",
    "Edad denunciada limpia",
    "Relación denunciante/denunciada",
    "Frecuencia de Episodios",
    "Nivel Educativo denunciante",
    "Condición Laboral denunciante",
    "Barrio hecho",
    "Localidad Hecho",
    "Departamento Hecho",
    "Origen",
    "Cantidad tipos violencia",
    "Gravedad máxima registrada",
    "Suma niveles violencia",
    "Cantidad medidas judiciales"
]

columnas_existentes = [col for col in columnas_casos if col in df.columns]

casos_powerbi = df[columnas_existentes].copy()

casos_powerbi["Relación categorizada"] = casos_powerbi["Relación denunciante/denunciada"].map(mapa_relacion)
casos_powerbi["Frecuencia categorizada"] = casos_powerbi["Frecuencia de Episodios"].map(mapa_frecuencia)
casos_powerbi["Nivel educativo denunciante categorizado"] = casos_powerbi["Nivel Educativo denunciante"].map(mapa_nivel_educativo)
casos_powerbi["Condición laboral denunciante categorizada"] = casos_powerbi["Condición Laboral denunciante"].map(mapa_condicion_laboral)
casos_powerbi["Gravedad máxima categorizada"] = casos_powerbi["Gravedad máxima registrada"].map(mapa_gravedad)

casos_powerbi.head()
```

```python
violencias_largo = df[["ID_Caso"] + variables_violencia].melt(
    id_vars="ID_Caso",
    value_vars=variables_violencia,
    var_name="Tipo de violencia",
    value_name="Nivel gravedad"
)

violencias_largo["Nivel gravedad"] = pd.to_numeric(
    violencias_largo["Nivel gravedad"],
    errors="coerce"
).fillna(0).astype(int)

violencias_largo["Violencia constatada"] = np.where(
    violencias_largo["Nivel gravedad"] > 0,
    "Sí",
    "No"
)

violencias_largo["Gravedad categorizada"] = violencias_largo["Nivel gravedad"].map(mapa_gravedad)

violencias_largo.head(12)
```

```python
violencias_constatadas = violencias_largo[
    violencias_largo["Nivel gravedad"] > 0
].copy()

violencias_constatadas.head()
```

```python
medidas_existentes = [col for col in medidas_judiciales if col in df.columns]

medidas_largo = df[["ID_Caso"] + medidas_existentes].melt(
    id_vars="ID_Caso",
    value_vars=medidas_existentes,
    var_name="Medida judicial",
    value_name="Aplicada"
)

medidas_largo["Aplicada"] = pd.to_numeric(
    medidas_largo["Aplicada"],
    errors="coerce"
).fillna(0).astype(int)

medidas_largo["Medida aplicada"] = np.where(
    medidas_largo["Aplicada"] > 0,
    "Sí",
    "No"
)

medidas_largo.head(12)
```

```python
medidas_aplicadas = medidas_largo[
    medidas_largo["Aplicada"] > 0
].copy()

medidas_aplicadas.head()
```

```python
calendario = casos_powerbi[["Fecha Intervención", "Año", "Mes"]].drop_duplicates().copy()

calendario = calendario.dropna(subset=["Fecha Intervención"])

calendario["Fecha Intervención"] = pd.to_datetime(calendario["Fecha Intervención"])

calendario["Nombre mes"] = calendario["Fecha Intervención"].dt.month_name(locale="Spanish")
calendario["Trimestre"] = "T" + calendario["Fecha Intervención"].dt.quarter.astype(str)

calendario = calendario.sort_values("Fecha Intervención")

calendario.head()
```

```python
kpis_generales = pd.DataFrame({
    "Indicador": [
        "Total de casos",
        "Total de casos con violencia psicológica",
        "Total de casos con violencia física",
        "Total de casos con violencia económica",
        "Total de casos con violencia sexual",
        "Total de casos con violencia ambiental",
        "Total de medidas judiciales aplicadas",
        "Promedio de tipos de violencia por caso",
        "Promedio de medidas judiciales por caso"
    ],
    "Valor": [
        len(df),
        (df["Violencia Psicológica"] > 0).sum(),
        (df["Violencia Física"] > 0).sum(),
        (df["Violencia Económica"] > 0).sum(),
        (df["Violencia Sexual"] > 0).sum(),
        (df["Violencia Ambiental"] > 0).sum(),
        df["Cantidad medidas judiciales"].sum(),
        round(df["Cantidad tipos violencia"].mean(), 2),
        round(df["Cantidad medidas judiciales"].mean(), 2)
    ]
})

kpis_generales
```

```python
with pd.ExcelWriter(RUTA_POWERBI) as writer:
    casos_powerbi.to_excel(writer, sheet_name="Casos", index=False)
    violencias_largo.to_excel(writer, sheet_name="Violencias largo", index=False)
    violencias_constatadas.to_excel(writer, sheet_name="Violencias constatadas", index=False)
    medidas_largo.to_excel(writer, sheet_name="Medidas largo", index=False)
    medidas_aplicadas.to_excel(writer, sheet_name="Medidas aplicadas", index=False)
    calendario.to_excel(writer, sheet_name="Calendario", index=False)
    kpis_generales.to_excel(writer, sheet_name="KPIs generales", index=False)

print("Archivo para Power BI guardado en:")
print(RUTA_POWERBI)
```
