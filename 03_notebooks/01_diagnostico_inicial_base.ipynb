```python
from pathlib import Path
import pandas as pd
import numpy as np

BASE_DIR = Path(r"C:\Users\Luciano Sotelo\Documents\TESIS_PODER_JUDICIAL")

RUTA_ORIGINAL = BASE_DIR / "01_datos_originales" / "base_poder_judicial_original_2022_2024.xlsx"
RUTA_LIMPIA = BASE_DIR / "02_datos_limpios" / "base_poder_judicial_limpia_v1.xlsx"

print(RUTA_ORIGINAL.exists())
```

```python
df = pd.read_excel(RUTA_ORIGINAL, sheet_name="2022-2023-2024")

print("Cantidad de filas:", df.shape[0])
print("Cantidad de columnas:", df.shape[1])

df.head()
```

```python
for i, col in enumerate(df.columns, start=1):
    print(i, col)
```

```python
resumen = pd.DataFrame({
    "columna": df.columns,
    "tipo_dato": df.dtypes.astype(str).values,
    "valores_nulos": df.isna().sum().values,
    "porcentaje_nulos": (df.isna().mean() * 100).round(2).values,
    "valores_unicos": df.nunique(dropna=True).values
})

resumen.sort_values("porcentaje_nulos", ascending=False).head(20)
```

```python
duplicados = df.duplicated().sum()
print("Filas duplicadas exactas:", duplicados)
```

```python
df_limpia = df.copy()

df_limpia.columns = (
    df_limpia.columns
    .str.strip()
    .str.replace("  ", " ", regex=False)
)

df_limpia = df_limpia.rename(columns={
    "sexo denunciante": "Sexo denunciante",
    "Fijacion de cuota alimentaria provisoria": "Fijación de cuota alimentaria provisoria",
    "Fijacion de regimen de visitas": "Fijación de régimen de visitas",
    "Guarda de menores/incpaces víctimas": "Guarda de menores/incapaces víctimas",
    "Reintegro al hogar a las vï¿½cticas": "Reintegro al hogar a las víctimas",
    "Intervenciones -Instit. No Gubernamentales": "Intervenciones - Instit. No Gubernamentales"
})

df_limpia.columns.tolist()
```

```python
def corregir_texto_mojibake(valor):
    if pd.isna(valor):
        return valor
    
    if not isinstance(valor, str):
        return valor
    
    texto = valor.strip()
    
    try:
        texto_corregido = texto.encode("latin1").decode("utf-8")
        return texto_corregido
    except:
        return texto

columnas_texto = [
    "Barrio hecho",
    "Localidad Hecho",
    "Departamento Hecho",
    "Origen"
]

for col in columnas_texto:
    if col in df_limpia.columns:
        df_limpia[col] = df_limpia[col].apply(corregir_texto_mojibake)

df_limpia[columnas_texto].head()
```

```python
df_limpia["Fecha Intervención"] = pd.to_datetime(
    df_limpia["Fecha Intervención"],
    errors="coerce",
    dayfirst=True
)

df_limpia["Año"] = df_limpia["Fecha Intervención"].dt.year
df_limpia["Mes"] = df_limpia["Fecha Intervención"].dt.month

df_limpia[["Fecha Intervención", "Año", "Mes"]].head()
```

```python
casos_por_anio = df_limpia["Año"].value_counts().sort_index()
casos_por_anio
```

```python
df_limpia["Edad denunciante limpia"] = pd.to_numeric(
    df_limpia["Edad denunciante"],
    errors="coerce"
)

df_limpia["Edad denunciada limpia"] = pd.to_numeric(
    df_limpia["Edad denunciada"],
    errors="coerce"
)

df_limpia.loc[
    (df_limpia["Edad denunciante limpia"] < 1) | 
    (df_limpia["Edad denunciante limpia"] > 100),
    "Edad denunciante limpia"
] = np.nan

df_limpia.loc[
    (df_limpia["Edad denunciada limpia"] < 1) | 
    (df_limpia["Edad denunciada limpia"] > 100),
    "Edad denunciada limpia"
] = np.nan

df_limpia[["Edad denunciante", "Edad denunciante limpia", "Edad denunciada", "Edad denunciada limpia"]].describe()
```

```python
variables_violencia = [
    "Violencia Física",
    "Violencia Psicológica",
    "Violencia Económica",
    "Violencia Sexual",
    "Violencia Restricción Participación Social",
    "Violencia Ambiental"
]

for col in variables_violencia:
    df_limpia[col] = pd.to_numeric(df_limpia[col], errors="coerce")
    df_limpia[col + " constatada"] = np.where(df_limpia[col] > 0, 1, 0)

df_limpia[[col + " constatada" for col in variables_violencia]].head()
```

```python
resumen_violencias = []

for col in variables_violencia:
    casos = df_limpia[col + " constatada"].sum()
    porcentaje = casos / len(df_limpia) * 100
    
    resumen_violencias.append({
        "Tipo de violencia": col,
        "Casos": int(casos),
        "Porcentaje": round(porcentaje, 2)
    })

resumen_violencias = pd.DataFrame(resumen_violencias)
resumen_violencias.sort_values("Casos", ascending=False)
```

```python
medidas_judiciales = [
    "Abstención de publicación en Redes Sociales",
    "Alojamiento temporario de la víctima",
    "Custodia policial",
    "Desestimación de denuncia",
    "Exclusión del agresor",
    "Fijación de cuota alimentaria provisoria",
    "Fijación de régimen de visitas",
    "Guarda de menores/incapaces",
    "Guarda de menores/incapaces víctimas",
    "Levantamiento de la medida judicial",
    "Oficio de protección policial",
    "Orden para retirar enseres personales",
    "Otras Medidas",
    "Prohibición de acercamiento o comunicación del agresor",
    "Recursos de Revocatoria",
    "Reintegro al hogar a las víctimas",
    "Restitución de alimentos",
    "Restitución de bienes",
    "Retiro de la víctima",
    "Secuestro de arma",
    "Tratamientos de rehabilitación"
]

resumen_medidas = []

for col in medidas_judiciales:
    if col in df_limpia.columns:
        df_limpia[col] = pd.to_numeric(df_limpia[col], errors="coerce").fillna(0)
        casos = df_limpia[col].sum()
        porcentaje = casos / len(df_limpia) * 100
        
        resumen_medidas.append({
            "Medida judicial": col,
            "Casos": int(casos),
            "Porcentaje": round(porcentaje, 2)
        })

resumen_medidas = pd.DataFrame(resumen_medidas)
resumen_medidas.sort_values("Casos", ascending=False).head(10)
```

```python
df_limpia.to_excel(RUTA_LIMPIA, index=False)

print("Base limpia guardada en:")
print(RUTA_LIMPIA)
```
